```python
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt
```

```python
dataset=load_dataset('lukebarousse/data_jobs')
df=dataset['train'].to_pandas()
```

```python
df['job_posted_date']=pd.to_datetime(df['job_posted_date'])
```

# counts of job Titles


```python
df.pivot_table(index='job_title_short', aggfunc='size')
```

```python
df.groupby('job_title_short').size()
```

```python
df.pivot_table(values='salary_year_avg',index='job_title_short',aggfunc='median')
```

```python
df.groupby('job_title_short')['salary_year_avg'].median()
```

```python
df.pivot_table(values='salary_year_avg',index='job_country',columns='job_title_short',aggfunc='median')
```

```python
median_salary_by_title = df.pivot_table(
    values='salary_year_avg',
    index='job_title_short',
    aggfunc='median'
)

median_salary_by_title = median_salary_by_title.sort_values(by='salary_year_avg', ascending=False)

plt.figure(figsize=(12, 8))
median_salary_by_title.plot(kind='bar', legend=False)
plt.title('Median Annual Salary by Job Title')
plt.ylabel('Median Salary (USD)')
plt.xlabel('Job Title')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
