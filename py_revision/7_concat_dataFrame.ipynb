# concatination we can add dataframes

```python
import pandas as pd
import matplotlib.pyplot as plt
from datasets import load_dataset
```

```python
dataset=load_dataset('lukebarousse/data_jobs')
df=dataset['train'].to_pandas()
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])

job_postings_jan = df[df['job_posted_date'].dt.month == 1]
job_postings_feb = df[df['job_posted_date'].dt.month == 2]
```

```python
df_q1 = pd.concat([job_postings_jan, job_postings_feb], ignore_index=True) # ignore_index=True creates new index
df_q1
```

```python
# make new month column
df['job_posted_month']=df['job_posted_date'].dt.strftime('%b')
df
```

```python
# list of month
months=df['job_posted_month'].unique()
```

```python
{month:month for month in months}
```

```python
dict_months = {month: df[df['job_posted_month'] == month] for month in months}
dict_months['Jan']
```

# BOSS needs 2023 quarter 1


```python
df_q1 = pd.concat([dict_months['Jan'], dict_months['Feb'], dict_months['Mar']], ignore_index=True)
if 'job_posted_month' not in df_q1.columns:
    df_q1['job_posted_month'] = pd.to_datetime(df_q1['job_posted_date']).dt.strftime('%b')
df_q1
```

```python
counts = df_q1['job_posted_month'].value_counts()

counts.plot(kind='bar')
plt.title('Job Postings by Month for Q1')
plt.xlabel('Month')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```
