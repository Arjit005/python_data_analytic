```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from datasets import load_dataset
```

```python
dataset=load_dataset('lukebarousse/data_jobs')
df=dataset['train'].to_pandas()
```

```python
fig,ax=plt.subplots(2,2)
```

```python
fig, ax = plt.subplots(1, 2, figsize=(14, 6))

# left: job schedule type (shows dominant Full-time)
vc_sched = df['job_schedule_type'].value_counts()
vc_sched.plot(kind='bar', ax=ax[0], color='steelblue')
ax[0].set_title('job_schedule_type')
ax[0].set_ylabel('count')
ax[0].tick_params(axis='x', rotation=45)

# right: top job titles
vc_titles = df['job_title_short'].value_counts().head(12)
vc_titles.plot(kind='bar', ax=ax[1], color='steelblue')
ax[1].set_title('job_title_short (top 12)')
ax[1].tick_params(axis='x', rotation=90)
ax[1].tick_params(labelsize=9)

fig.tight_layout()
```

```python
fig, ax = plt.subplots(figsize=(6,6))
df['job_work_from_home'].value_counts().plot(kind='pie', autopct='%1.1f%%', ax=ax, colors=['steelblue','lightgray'])
ax.set_ylabel('')
ax.set_title('Work From Home')
plt.tight_layout()
plt.show()
```
