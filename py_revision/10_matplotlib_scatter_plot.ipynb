```python
import pandas as pd
import matplotlib.pyplot as plt

# Sample Data
data = {
    'skill_count': [93000, 58000, 46000, 67000, 39500, 30000, 28000, 11000, 13500, 13000],
    'median_salary': [92500, 98500, 95000, 84500, 90000, 92500, 90000, 90000, 85000, 80800]
}

# Skills as index
skills = [
    'sql', 'python', 'tableau', 'excel',
    'power bi', 'r', 'sas', 'sap',
    'powerpoint', 'word'
]

skills_stats = pd.DataFrame(data, index=skills)
```

```python
# Scatter Plot
fig, ax = plt.subplots(figsize=(12, 7))
skills_stats.plot(
    kind='scatter',
    x='skill_count',
    y='median_salary',
    ax=ax,
    s=120,
    alpha=0.7,
    color='steelblue'
)

ax.set_xlabel('Count of Job Postings', fontsize=12, fontweight='bold')
ax.set_ylabel('Median Yearly Salary ($USD)', fontsize=12, fontweight='bold')
ax.set_title('Salary vs. Count of Job Postings for Top 10 Skills', fontsize=14, fontweight='bold')

# Add labels to each point
for i, txt in enumerate(skills_stats.index):
    x = skills_stats['skill_count'].iloc[i]
    y = skills_stats['median_salary'].iloc[i]
    ax.annotate(
        txt,
        xy=(x, y),
        xytext=(8, 8),
        textcoords='offset points',
        fontsize=9,
        fontweight='bold',
        bbox=dict(boxstyle='round,pad=0.25', facecolor='white', alpha=0.8)
    )

ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from adjustText import adjust_text

# Data
data = {
    'skill': [
        'sql', 'python', 'excel', 'tableau', 'power bi',
        'r', 'sas', 'powerpoint', 'word', 'sap',
        'go', 'vba', 'sql server', 'oracle', 'flow',
        'azure', 'aws', 'qlik', 'looker', 'snowflake'
    ],
    'skill_count': [
        93000, 58000, 67000, 50000, 39500,
        30000, 28000, 15000, 15000, 14000,
        9000, 10000, 11000, 12000, 9000,
        12000, 11000, 8500, 8000, 7800
    ],
    'median_salary': [
        92500, 98500, 84500, 95000, 90000,
        92500, 90000, 85000, 80800, 90000,
        90000, 92500, 94500, 95200, 95000,
        100000, 100500, 99500, 104000, 111000
    ]
}

skills_stats = pd.DataFrame(data).set_index('skill')

# Plot
ax = skills_stats.plot(
    kind='scatter',
    x='skill_count',
    y='median_salary',
    figsize=(10, 7),
    s=40
)

# Add labels
texts = []

for i, txt in enumerate(skills_stats.index):
    texts.append(
        plt.text(
            skills_stats['skill_count'].iloc[i],
            skills_stats['median_salary'].iloc[i],
            txt,
            fontsize=10
        )
    )

# Prevent overlapping labels
adjust_text(
    texts,
    arrowprops=dict(
        arrowstyle='->',
        color='gray'
    )
)

# Format salary axis
ax.yaxis.set_major_formatter(
    plt.FuncFormatter(lambda y, pos: f'${int(y/1000)}K')
)

# Labels and title
plt.xlabel('Count of Job Postings', fontsize=12)
plt.ylabel('Median Yearly Salary', fontsize=12)
plt.title('Salary vs. Count of Job Postings for Top 20 Skills', fontsize=14)

plt.tight_layout()
plt.show()
```
