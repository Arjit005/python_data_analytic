# sort_index()

```python
df_usa = df.loc[df['job_country'] == 'United States'].copy()
median_pivot = df_usa.pivot_table(
    values='salary_year_avg',
    index='job_title_short',
    aggfunc=['median', 'max','min']
)
median_pivot
median_pivot
```

```python
median_pivot = median_pivot.sort_values(by=('median', 'salary_year_avg'), ascending=False)
median_pivot
```

```python
median_pivot.sort_index(by='job_title_short',ascending=False)
```
