# Investigate trending skills

```python
import ast
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

#
dataset=load_dataset('lukebarousse/data_jobs')
df=dataset['train'].to_pandas()
#data clean up
df['job_posted_date']=pd.to_datetime(df['job_posted_date'])
df['job_skills']=df['job_skills'].apply(lambda x: ast.literal_eval(x) if pd.notna(x) else x)
```

```python
df_DA=df[df['job_title_short']== 'Data Analyst'].copy()
```

```python
df_DA
```

```python
df_DA['job_posted_month_no']=df_DA['job_posted_date'].dt.month # new column of month
df_DA
```

```python
df_DA_explode = df_DA.explode('job_skills').rename(columns={'job_skills': 'job_skill'})
```

```python
df_DA_explode
```

```python
df_DA_explode.pivot_table(index='job_posted_month_no', columns='job_skill', aggfunc='size',fill_value=0)
```

```python
df_DA_pivot = df_DA_explode.pivot_table(index='job_posted_month_no', columns='job_skill', aggfunc='size', fill_value=0)
df_DA_pivot.plot(kind='line')
```

```python
df_DA_pivot.loc['Total']=df_DA_pivot.sum()
```

```python
df_DA_pivot.loc['Total'].sort_values(ascending=False)
```

```python
df_DA_pivot[df_DA_pivot.loc['Total'].sort_values(ascending=False).index]
```

```python
df_DA_pivot.iloc[:,:5].plot(kind='line')
plt.ylabel('count')
plt.show()
```
