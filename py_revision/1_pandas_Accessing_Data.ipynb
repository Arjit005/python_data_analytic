# Data cleaning


```python
df['job_posted_date']=pd.to_datetime(df['job_posted_date']) # cahnging type from obj to datetime
```

```python
# fill null values with median
median_salary_year=df['salary_year_avg'].median()

median_salary_hour=df['salary_hour_avg'].median()
```

```python
# we create new data frame to keep old dataframe safe
df_filled=df

df_filled['salary_year_avg']=df_filled['salary_year_avg'].fillna(median_salary_year) # filling null values

df_filled['salary_hour_avg']=df_filled['salary_hour_avg'].fillna(median_salary_hour)
```

```python
df.loc[:,'salary_rate':'salary_hour_avg'] # seeing data
```

```python
# remove duplicates
df_unique=df_filled # creating new data frame
df_unique.drop_duplicates()# now removed until we make equal to df_unique
df_unique=df_unique.drop_duplicates()
```

```python
print('lenght of original df',len(df_filled))
print('lenght of drop_duplicate df',len(df_unique))
print('Row_dropped',len(df_filled)-len(df_unique))


```
